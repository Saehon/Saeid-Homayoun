```python
from pathlib import Path
import zipfile
root = Path('/content/Saeid-Homayoun/studies/FF-Damodaran-Discovery-01')
out = Path('/content/ff-damodaran-crosswalk-review.zip')
files = [
    root/'crosswalk'/'candidate_crosswalk.csv',
    root/'evidence'/'download_manifest.json',
    root/'evidence'/'data_passport.json',
]
with zipfile.ZipFile(out, 'w', zipfile.ZIP_DEFLATED) as zf:
    for p in files:
        if p.exists():
            zf.write(p, arcname=p.relative_to(root))
print(out)
```
